# Aviation Accidents Analysis

You are part of a consulting firm that is tasked to do an analysis of commercial and passenger jet airline safety. The client (an airline/airplane insurer) is interested in knowing what types of aircraft (makes/models) exhibit low rates of total destruction and low likelihood of fatal or serious passenger injuries in the event of an accident. They are also interested in any general variables/conditions that might be at play. Your analysis will be based off of aviation accident data accumulated from the years 1948-2023. 

Our client is only interested in airplane makes/models that are professional builds and could potentially still be active. Assume a max lifetime of 40 years for a make/model retirement and make sure to filter your data accordingly (i.e. from 1983 onwards). They would also like separate recommendations for small aircraft vs. larger passenger models. **In addition, make sure that claims that you make are statistically robust and that you have enough samples when making comparisons between groups.**

- **conduct EDA: visualization and statistical measures to understand the structure of the data**

```python
# loading relevant packages
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

clean_df = pd.read_csv("cleaned_aviation_data.csv")

clean_df.head()
clean_df.info()
```

## Exploratory Data Analysis  
- Load in the cleaned data

```python
clean_df['Event_Date'] = pd.to_datetime(clean_df['Event_Date'], errors='coerce')
active_df = clean_df[clean_df['Event_Date'].dt.year >= 1983]

print("Active aircraft accidents:", active_df.shape)
```

## Explore safety metrics across models/makes
- Remember that the client is interested in separate recommendations for smaller airplanes and larger airplanes. Choose a passenger threshold of 20 and separate the plane types. 

```python
active_df.columns = active_df.columns.str.strip().str.replace(' ', '_').str.replace('.', '_')

numeric_cols = ['Total_Fatal_Injuries', 'Total_Serious_Injuries', 'Destroyed']
for col in numeric_cols:
    active_df[col] = pd.to_numeric(active_df[col], errors='coerce').fillna(0)
```

#### Analyzing Makes

Explore the human injury risk profile for small and larger Makes:
- choose the 15 makes for each group possessing the lowest mean fatal/seriously injured fraction
- plot the mean fatal/seriously injured fraction for each of these subgroups side-by-side

```python
small_aircraft = active_df[active_df['Aircraft_Category'].str.lower().str.contains('small', na=False)]
large_aircraft = active_df[~active_df['Aircraft_Category'].str.lower().str.contains('small', na=False)]

print("Small aircraft sample:", small_aircraft.shape)
print("Large aircraft sample:", large_aircraft.shape)
```

**Distribution of injury rates: small makes**

Use a violinplot to look at the distribution of the fraction of passengers serious/fatally injured for small airplane makes. Just display makes with the ten lowest mean serious/fatal injury rates.

```python
def summarize_safety(df):
    return df.groupby(['Make','Model']).agg(
        Accidents=('Event_Id','count'),
        Avg_Fatal_Injuries=('Total_Fatal_Injuries','mean'),
        Avg_Serious_Injuries=('Total_Serious_Injuries','mean'),
        Destruction_Rate=('Destroyed','mean')
    ).reset_index().sort_values('Destruction_Rate')

small_summary = summarize_safety(small_aircraft)
large_summary = summarize_safety(large_aircraft)
```

**Distribution of injury rates: large makes**

Use a stripplot to look at the distribution of the fraction of passengers serious/fatally injured for large airplane makes. Just display makes with the ten lowest mean serious/fatal injury rates.

```python
small_summary['Serious_Fatal_Fraction'] = small_summary['Avg_Fatal_Injuries'] + small_summary['Avg_Serious_Injuries']
top_small = small_summary.nsmallest(10, 'Serious_Fatal_Fraction')

sns.barplot(data=top_small, x='Model', y='Serious_Fatal_Fraction')
plt.xticks(rotation=45)
plt.title("Top 10 Safest Small Aircraft Models")
plt.ylabel("Mean Serious + Fatal Injuries")
plt.xlabel("Aircraft Model")
plt.tight_layout()
plt.show()
```

**Evaluate the rate of aircraft destruction for both small and large aircraft by Make.** 

Sort your results and keep the lowest 15.

```python
large_summary['Serious_Fatal_Fraction'] = large_summary['Avg_Fatal_Injuries'] + large_summary['Avg_Serious_Injuries']
top_large = large_summary.nsmallest(10, 'Serious_Fatal_Fraction')

sns.barplot(data=top_large, x='Model', y='Serious_Fatal_Fraction')
plt.xticks(rotation=45)
plt.title("Top 10 Safest Large Aircraft Models")
plt.ylabel("Mean Serious + Fatal Injuries")
plt.xlabel("Aircraft Model")
plt.tight_layout()
plt.show()
```

#### Provide a short discussion on your findings for your summary statistics and plots:
- Make any recommendations for Makes here based off of the destroyed fraction and fraction fatally/seriously injured
- Comment on the calculated statistics and any corresponding distributions you have visualized.

```python
phase_summary = (
    active_df.groupby('Broad_phase_of_flight')
    .agg(
        Destruction_Rate=('Destroyed', 'mean'),
        Avg_Serious_Fatal=('Total_Fatal_Injuries', 'sum')
    )
    .assign(Avg_Serious_Fatal=lambda df: df['Avg_Serious_Fatal'] + active_df.groupby('Broad_phase_of_flight')['Total_Serious_Injuries'].sum())
    .reset_index()
)

plt.figure(figsize=(10,6))
sns.barplot(data=phase_summary, x='Broad_phase_of_flight', y='Destruction_Rate')
plt.xticks(rotation=45)
plt.title("Destruction Rate by Phase of Flight")
plt.tight_layout()
plt.show()
```

### Analyze plane types
- plot the mean fatal/seriously injured fraction for both small and larger planes 
- also provide a distributional plot of your choice for the fatal/seriously injured fraction by airplane type (stripplot, violin, etc)  
- filter ensuring that you have at least ten individual examples in each model/make to average over

**Larger planes**

```python
weather_summary = (
    active_df.groupby('Weather_Condition')
    .agg(
        Destruction_Rate=('Destroyed', 'mean'),
        Total_Fatal=('Total_Fatal_Injuries', 'sum'),
        Total_Serious=('Total_Serious_Injuries', 'sum')
    )
    .assign(Avg_Serious_Fatal=lambda df: df['Total_Fatal'] + df['Total_Serious'])
    .reset_index()
)

plt.figure(figsize=(10,6))
sns.barplot(data=weather_summary, x='Weather_Condition', y='Destruction_Rate')
plt.xticks(rotation=45)
plt.title("Destruction Rate by Weather Condition")
plt.tight_layout()
plt.show()
```

**Smaller planes**
- for smaller planes, limit your plotted results to the makes with the 10 lowest mean serious/fatal injury fractions

```python
engine_summary = (
    active_df.groupby('Engine_Type')
    .agg(
        Destruction_Rate=('Destroyed', 'mean'),
        Total_Fatal=('Total_Fatal_Injuries', 'sum'),
        Total_Serious=('Total_Serious_Injuries', 'sum')
    )
    .assign(Avg_Serious_Fatal=lambda df: df['Total_Fatal'] + df['Total_Serious'])
    .reset_index()
)

plt.figure(figsize=(10,6))
sns.barplot(data=engine_summary, x='Engine_Type', y='Destruction_Rate')
plt.xticks(rotation=45)
plt.title("Destruction Rate by Engine Type")
plt.tight_layout()
plt.show()
```
